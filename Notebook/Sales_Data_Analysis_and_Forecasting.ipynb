# Phase 4 — Modeling

## Step 1: Prepare Monthly Sales Data

The transaction-level sales data is aggregated into monthly sales
to prepare the dataset for time-based forecasting.

```python
# Prepare monthly sales data

monthly_sales = (
    df_clean
    .groupby(df_clean["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
    .reset_index()
)

monthly_sales["Order Date"] = monthly_sales["Order Date"].dt.to_timestamp()

monthly_sales.columns = ["Date", "Sales"]

print(monthly_sales.head())
print("\nShape:", monthly_sales.shape)
```

```python
# Create time-based features

monthly_sales["Year"] = monthly_sales["Date"].dt.year
monthly_sales["Month"] = monthly_sales["Date"].dt.month
monthly_sales["Time_Index"] = np.arange(len(monthly_sales))

print(monthly_sales.head())
```

```python
# Create time-based train-test split

split_index = int(len(monthly_sales) * 0.8)

train = monthly_sales.iloc[:split_index].copy()
test = monthly_sales.iloc[split_index:].copy()

print("Training data:", train.shape)
print("Testing data:", test.shape)

print("\nTraining period:")
print(train["Date"].min(), "to", train["Date"].max())

print("\nTesting period:")
print(test["Date"].min(), "to", test["Date"].max())
```

```python
# Prepare features and target

features = ["Time_Index", "Year", "Month"]

X_train = train[features]
y_train = train["Sales"]

X_test = test[features]
y_test = test["Sales"]

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Import Linear Regression

from sklearn.linear_model import LinearRegression

# Create the model
model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
# Predict sales for the test period

y_pred = model.predict(X_test)

# Display actual vs predicted values

results = pd.DataFrame({
    "Date": test["Date"],
    "Actual Sales": y_test.values,
    "Predicted Sales": y_pred
})

results
```

```python
# Evaluate the forecasting model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation")
print("------------------------")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
# Actual vs Predicted Sales

plt.figure(figsize=(12, 6))

plt.plot(
    results["Date"],
    results["Actual Sales"],
    marker="o",
    label="Actual Sales"
)

plt.plot(
    results["Date"],
    results["Predicted Sales"],
    marker="o",
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```

### Model Evaluation Insight

The Linear Regression model captures the general upward trend in monthly
sales but does not capture the strong month-to-month fluctuations in the
actual sales. The model underestimates several high-sales months and
overestimates some lower-sales months. This is consistent with the relatively
low R² score of 0.153, indicating that the baseline model has limited
predictive accuracy.

```python
# Random Forest Regression Model

from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Random Forest Predictions

rf_pred = rf_model.predict(X_test)

rf_results = pd.DataFrame({
    "Date": test["Date"],
    "Actual Sales": y_test.values,
    "Predicted Sales": rf_pred
})

rf_results
```

```python
# Evaluate Random Forest Model

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_mse = mean_squared_error(y_test, rf_pred)
rf_rmse = np.sqrt(rf_mse)
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Model Evaluation")
print("--------------------------------")
print("MAE :", rf_mae)
print("MSE :", rf_mse)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)
```

## Model Comparison and Selection

Two regression models were evaluated for monthly sales forecasting:
Linear Regression and Random Forest Regression.

Linear Regression achieved an MAE of approximately 51,799, RMSE of
approximately 62,485, and an R² score of 0.153.

Random Forest achieved an MAE of approximately 78,192, RMSE of
approximately 90,177, and an R² score of -0.763.

Based on the evaluation metrics, Linear Regression performed better
than Random Forest and was selected as the final model.

```python
# Create lag features for time-series forecasting

lag_data = monthly_sales[["Date", "Sales"]].copy()

lag_data["Lag_1"] = lag_data["Sales"].shift(1)
lag_data["Lag_2"] = lag_data["Sales"].shift(2)
lag_data["Lag_12"] = lag_data["Sales"].shift(12)

# Remove rows with missing lag values
lag_data = lag_data.dropna().reset_index(drop=True)

print(lag_data.head())
print("\nShape:", lag_data.shape)
```

```python
# Create time-based train-test split for lag model

split_index = int(len(lag_data) * 0.8)

lag_train = lag_data.iloc[:split_index].copy()
lag_test = lag_data.iloc[split_index:].copy()

print("Training data:", lag_train.shape)
print("Testing data:", lag_test.shape)

print("\nTraining period:")
print(lag_train["Date"].min(), "to", lag_train["Date"].max())

print("\nTesting period:")
print(lag_test["Date"].min(), "to", lag_test["Date"].max())
```

```python
# Prepare lag features and target

lag_features = ["Lag_1", "Lag_2", "Lag_12"]

X_lag_train = lag_train[lag_features]
y_lag_train = lag_train["Sales"]

X_lag_test = lag_test[lag_features]
y_lag_test = lag_test["Sales"]

print("X_train shape:", X_lag_train.shape)
print("X_test shape:", X_lag_test.shape)
```

```python
# Train Linear Regression with lag features

lag_model = LinearRegression()

lag_model.fit(X_lag_train, y_lag_train)

print("Improved Linear Regression model trained successfully!")
```

```python
# Make predictions using lag features

lag_pred = lag_model.predict(X_lag_test)

lag_results = pd.DataFrame({
    "Date": lag_test["Date"],
    "Actual Sales": y_lag_test.values,
    "Predicted Sales": lag_pred
})

lag_results
```

```python
# Evaluate improved Linear Regression model

lag_mae = mean_absolute_error(y_lag_test, lag_pred)
lag_mse = mean_squared_error(y_lag_test, lag_pred)
lag_rmse = np.sqrt(lag_mse)
lag_r2 = r2_score(y_lag_test, lag_pred)

print("Improved Linear Regression Evaluation")
print("---------------------------------------")
print("MAE :", lag_mae)
print("MSE :", lag_mse)
print("RMSE:", lag_rmse)
print("R²  :", lag_r2)
```

```python

# Actual vs Predicted Sales - Final Model

plt.figure(figsize=(12, 6))

plt.plot(
    lag_results["Date"],
    lag_results["Actual Sales"],
    marker="o",
    label="Actual Sales"
)

plt.plot(
    lag_results["Date"],
    lag_results["Predicted Sales"],
    marker="o",
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Monthly Sales - Final Model")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```
